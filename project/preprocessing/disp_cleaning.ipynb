```python
import pandas as pd
```

```python
disp=pd.read_csv(r"disp.csv",sep=";")
disp.head(5)
```

```python
disp.info()
```

```python
disp["disp_id"] = disp["disp_id"].astype(str)
disp["client_id"] = disp["client_id"].astype(str)
disp["account_id"] = disp["account_id"].astype(str)
```

```python
disp["type"] = disp["type"].str.upper().str.strip()
```

## Final Check

```python
print(disp.info())
print("\nMissing Values:\n", disp.isnull().sum())
print("\nUnique Dispositions:", disp["disp_id"].nunique())
print("\nUnique Clients:", disp["client_id"].nunique())
print("\nUnique Accounts:", disp["account_id"].nunique())
print("\nDuplicates:", disp.duplicated().sum())
print("\nRole Distribution:\n", disp["type"].value_counts())
disp.head()
```
